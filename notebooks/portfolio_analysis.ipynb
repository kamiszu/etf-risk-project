```python
import pandas as pd
```

```python
prices = pd.read_csv("../data/raw/prices.csv", index_col="Date")
prices.index = pd.to_datetime(prices.index)
prices.head()
```

```python
returns = prices.pct_change()
```

```python
returns.dropna()
```

```python
portfolio = pd.read_csv("../data/portfolio.csv")
portfolio
```

```python
weights = dict(zip(portfolio.ticker, portfolio.weight))
weights
```

```python
returns = returns[portfolio.ticker]
portfolio_returns = returns.mul(portfolio.weight.values, axis=1).sum(axis=1)
portfolio_returns.head()
```
